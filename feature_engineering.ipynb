```python
import seaborn as sns
import matplotlib.pyplot as plt

# -------------------------------------
# 1. Обчислюємо кореляційну матрицю один раз
# -------------------------------------
corr_matrix = df.corr()

target = "close"
corr_target = corr_matrix[target].sort_values(ascending=False)

print("\nTop correlated features with close:")
print(corr_target.head(100))
```
