```python
import pandas as pd
df = pd.read_csv("data/btc_hourly_5y_feature.csv")
df.head()
```

```python


# беремо тільки числові фічі
numeric_df = df.select_dtypes(include=["float64", "int64"])

# витягуємо кореляції з target
target_corr = numeric_df.corr()["target"].sort_values(ascending=False)

print(target_corr)
```

```python
# топ 10 позитивних
print(target_corr.head(10))

# топ 10 негативних
print(target_corr.tail(10))
```
